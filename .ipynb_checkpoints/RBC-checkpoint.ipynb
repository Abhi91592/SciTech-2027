```python
from CoolProp.CoolProp import PropsSI

test_density = PropsSI('D', 'T', 300, 'P', 1e5, 'CarbonDioxide')
print(test_density)
```

```python
# User Input, Heat Exchanger output (FIXED)
T_HX_out = 509.6      # Kelvin
P_HX_in_guess = 9.338 # MPa

# Convergence tolerance
TOL_T = 0.01  # K
TOL_P = 0.01  # Pa

max_iter = 5000
iter = 0
```

```python
while True:
    iter += 1

    # Define physical bounds for CO2 conservative
    T_min = 270       # Minimum temperature [K]
    T_max = 1500      # Maximum temperature [K]
    P_min = 1e5       # Minimum pressure [Pa]
    P_max = 30e7      # Maximum pressure [Pa]

    # Initial guesses
    T_HX_in_guess = 400       # K

    T_comp_out_guess = 300    # K
    P_comp_out_guess = 9.432  # MPa

    # Use guesses for first iteration, then previous outputs
    if iter == 1:
        Re_T_cold_in = T_comp_out_guess
        Re_P_cold_in = P_comp_out_guess * 1e5  # kept same as MATLAB
        T_HX_in = T_HX_in_guess
        P_HX_in = P_HX_in_guess * 1e5
    else:
        Re_T_cold_in = prev_T_comp_out_actual
        Re_P_cold_in = prev_P_comp_out
        T_HX_in = Re_T_cold_out
        P_HX_in = Re_P_cold_out

    if iter > max_iter:
        warnings.warn("Did not converge within max iterations")
        break

    # Heat Input ThorCon Summary
    Q = 220000  # Output Power from the heater 220 kW

# HX
    T_HX_in = T_HX_in_guess
    P_HX_in_MPa = P_HX_in_guess
    P_HX_in = P_HX_in_MPa * 1e6
    P_HX_Drop = 0.00
    P_HX_out = (P_HX_in * 1 - P_HX_Drop)
    T_HX_Avg = (T_HX_in + T_HX_out) / 2
    P_HX_Avg = (P_HX_out + P_HX_in) / 2
    e_HX = 1.0
# HEX Calculations
    cph = PropsSI(
        'CPMASS',
        'T', 423.599,
        'P', ((20000000 - 3000 + 20000000) / 2),
        'CarbonDioxide'
    )

    h_sCO2_in_HX = PropsSI(
        'H',
        'T', T_HX_in,
        'P', P_HX_in,
        'CarbonDioxide'
    )

    h_sCO2_out_HX = PropsSI(
        'H',
        'T', T_HX_out,
        'P', P_HX_out,
        'CarbonDioxide'
    )

    s_sCO2_in_HX = PropsSI(
        'S',
        'T', T_HX_in,
        'P', P_HX_in,
        'CarbonDioxide'
    )

    s_sCO2_out_HX = PropsSI(
        'S',
        'T', T_HX_out,
        'P', P_HX_in,
        'CarbonDioxide'
    )

    sco2_m_flow = (e_HX * Q) / (h_sCO2_out_HX - h_sCO2_in_HX)
# Turbine
    T_t_in = T_HX_out
    P_t_in = P_HX_out
    T_pressure_ratio = 1.198562444
    P_t_out = P_t_in / T_pressure_ratio
    e_Turbine = 0.84
    e_gearbox = 0.99
    e_generator = 0.985

    # Turbine Inlet Calculations
    s_in_turbine = PropsSI(
        'S',
        'T', T_t_in,
        'P', P_t_in,
        'CarbonDioxide'
    )

    h_sCO2_in_turbine = PropsSI(
        'H',
        'T', T_t_in,
        'P', P_t_in,
        'CarbonDioxide'
    ) / 1000

    # Turbine Isentropic Outlet Calculations
    T_t_out_isentropic = PropsSI(
        'T',
        'P', P_t_out,
        'S', s_in_turbine,
        'CarbonDioxide'
    )

    s_out_turbine_Isentropic = PropsSI(
        'S',
        'T', T_t_out_isentropic,
        'P', P_t_out,
        'CarbonDioxide'
    )

    h_sCO2_out_turbine_Isentropic = PropsSI(
        'H',
        'T', T_t_out_isentropic,
        'P', P_t_out,
        'CarbonDioxide'
    ) / 1000

    # Turbine Actual Outlet Calculations
    T_t_out_actual = T_t_in - e_Turbine * (T_t_in - T_t_out_isentropic)

    s_out_turbine = PropsSI(
        'S',
        'T', T_t_out_actual,
        'P', P_t_out,
        'CarbonDioxide'
    )

    h_sCO2_out_turbine = PropsSI(
        'H',
        'T', T_t_out_actual,
        'P', P_t_out,
        'CarbonDioxide'
    ) / 1000

    # Turbine Power Output Calculations
    h_Turbine_change = h_sCO2_in_turbine - h_sCO2_out_turbine
    Power_Output = sco2_m_flow * h_Turbine_change
    Power_Output_e = Power_Output * e_generator * e_gearbox

    # Gamma Calculations
    T_t_avg = (T_t_in + T_t_out_actual) / 2
    P_t_avg = (P_t_in + P_t_out) / 2

    T_cp = PropsSI(
        'CPMASS',
        'T', T_t_avg,
        'P', P_t_avg,
        'CarbonDioxide'
    )

    T_cv = PropsSI(
        'CVMASS',
        'T', T_t_avg,
        'P', P_t_avg,
        'CarbonDioxide'
    )

    T_gamma = T_cp / T_cv
# Recuperator
    if iter == 1:
        Re_T_cold_in = T_comp_out_guess
    else:
        Re_T_cold_in = T_comp_out_actual

    if iter == 1:
        Re_P_cold_in = P_comp_out_guess * 1e5
    else:
        Re_P_cold_in = P_comp_out

    Re_P_Drop = 0.01
    Re_T_hot_in = T_t_out_actual
    Re_P_hot_in = P_t_out
    Re_P_hot_out = Re_P_hot_in * (1 - Re_P_Drop)
    Re_P_cold_out = Re_P_cold_in * (1 - Re_P_Drop)

    e_Recup = 0.95

    # Enthalpies at recuperator inlets
    Re_h_hot_in = PropsSI(
        'H',
        'T', Re_T_hot_in,
        'P', Re_P_hot_in,
        'CarbonDioxide'
    )

    Re_h_cold_in = PropsSI(
        'H',
        'T', Re_T_cold_in,
        'P', Re_P_cold_in,
        'CarbonDioxide'
    )

    # Assume recuperator hot side outlet temperature
    Re_T_hot_out = Re_T_hot_in - e_Recup * (Re_T_hot_in - Re_T_cold_in)

    # Enthalpies at recuperator outlet
    Re_h_hot_out = PropsSI(
        'H',
        'T', Re_T_hot_out,
        'P', Re_P_hot_out,
        'CarbonDioxide'
    )

    # Actual recuperator heat transfer
    Q_dot_recup = sco2_m_flow * (Re_h_hot_in - Re_h_hot_out)

    # Cold outlet enthalpy
    h_cold_out = Re_h_cold_in + Q_dot_recup / sco2_m_flow

    # Cold outlet temperature
    Re_T_cold_out = PropsSI(
        'T',
        'H', h_cold_out,
        'P', Re_P_cold_in,
        'CO2'
    )

    # Entropy Calculations
    Re_s_hot_in = PropsSI(
        'S',
        'T', Re_T_hot_in,
        'P', Re_P_hot_in,
        'CarbonDioxide'
    )

    Re_s_hot_out = PropsSI(
        'S',
        'T', Re_T_hot_out,
        'P', Re_P_hot_out,
        'CarbonDioxide'
    )

    Re_s_cold_in = PropsSI(
        'S',
        'T', Re_T_cold_in,
        'P', Re_P_cold_in,
        'CarbonDioxide'
    )

    Re_s_cold_out = PropsSI(
        'S',
        'T', Re_T_cold_out,
        'P', Re_P_cold_out,
        'CarbonDioxide'
    )
# Cooler
    T_cooler_in = Re_T_hot_out
    cooler_Pressure_loss = 0.02
    P_cooler_in = Re_P_hot_out
    P_cooler_out = (1 - cooler_Pressure_loss) * P_cooler_in
    e_intercooler = 0.80
    T_cooler_amb = 295

    # Intercooler Temperature and Enthalpy Calculations
    T_cooler_out = T_cooler_in - e_intercooler * (T_cooler_in - T_cooler_amb)

    h_cooler_in = PropsSI(
        'H',
        'T', T_cooler_in,
        'P', P_cooler_in,
        'CarbonDioxide'
    ) / 1000

    h_cooler_out = PropsSI(
        'H',
        'T', T_cooler_out,
        'P', P_cooler_out,
        'CarbonDioxide'
    ) / 1000

    s_cooler_in = PropsSI(
        'S',
        'T', T_cooler_in,
        'P', P_cooler_in,
        'CarbonDioxide'
    ) / 1000

    s_cooler_out = PropsSI(
        'S',
        'T', T_cooler_out,
        'P', P_cooler_out,
        'CarbonDioxide'
    ) / 1000

    # Heat Rejected in the intercooler
    cooler_Q_reject = sco2_m_flow * (h_cooler_in - h_cooler_out)
# Compressor
    T_comp_in = T_cooler_out
    P_comp_in = P_cooler_out
    e_comp = 0.66
    p_ratio_comp = 1.23747048
# Compressor Calculations
    s_comp_in = float(PropsSI(
        'S',
        'T', T_comp_in,
        'P', P_comp_in,
        'CarbonDioxide'
    ))

    h_comp_in = float(PropsSI(
        'H',
        'T', T_comp_in,
        'P', P_comp_in,
        'CarbonDioxide'
    ))

    P_comp_out = P_comp_in * p_ratio_comp

    h_comp_out_isentropic = float(PropsSI(
        'H',
        'P', P_comp_out,
        'S', s_comp_in,
        'CarbonDioxide'
    ))

    h_comp_out = h_comp_in + (h_comp_out_isentropic - h_comp_in) / e_comp

    T_comp_out_actual = float(PropsSI(
        'T',
        'P', P_comp_out,
        'H', h_comp_out,
        'CarbonDioxide'
    ))

    comp_power = sco2_m_flow * (h_comp_out - h_comp_in) / 1e3

    s_comp_out_isentropic = PropsSI(
        'S',
        'S', s_comp_in,
        'P', P_comp_out,
        'CarbonDioxide'
    )

    s_comp_out = PropsSI(
        'S',
        'T', T_comp_out_actual,
        'P', P_comp_out,
        'CarbonDioxide'
    )

    T_comp_out_isentropic = float(PropsSI(
        'T',
        'P', P_comp_out,
        'H', h_comp_out_isentropic,
        'CarbonDioxide'
    ))

    # Update guess and check convergence
    Re_T_cold_out = float(Re_T_cold_out)
    Re_P_cold_out_val = float(Re_P_cold_out)

    # Always update compressor outlet guesses
    Re_T_cold_in = T_comp_out_actual
    Re_P_cold_in = P_comp_out / 1e5

    # Check for convergence
    prev_T_comp_out_actual = T_comp_out_actual
    prev_P_comp_out = P_comp_out

    delta_T = abs(T_HX_in_guess - Re_T_cold_out)
    delta_P = abs(P_HX_in_guess - Re_P_cold_out_val / 1e5)
    delta_T_comp = abs(T_comp_out_guess - T_comp_out_actual)

    # Final convergence check
    if delta_T < TOL_T and delta_P < TOL_P and delta_T_comp < TOL_T:
        break
```

```python
# --------------------------
# Printout Statements
# --------------------------

print("\nHEAT INPUT FROM TERTIARY SALT")
print(f"\nOutput from the Heater                   = {Q:.2f} W")

# HEX Output Statements
print(f"\nHeat Exchanger Inlet Temperature         = {T_HX_in:.2f} K")
print(f"Heat Exchanger Outlet Temperature        = {T_HX_out:.2f} K")
print(f"Heat Exchanger Inlet Pressure            = {P_HX_in:.2f} Pa")
print(f"Heat Exchanger Outlet Pressure           = {P_HX_out:.2f} Pa")
print(f"Heat Exchanger Average Pressure          = {P_HX_Avg:.2f} Pa")
print(f"Heat Exchanger Average Temperature       = {T_HX_Avg:.2f} K")
print(f"Heat Exchanger Specific Heat             = {cph:.2f} J/kg")
print(f"Heat Exchanger Mass flow rate of sCO2    = {sco2_m_flow:.4f} kg/s")
print(f"Heat Exchanger Inlet Enthalpy            = {h_sCO2_in_HX/1000:.2f} KJ/kg")
print(f"Heat Exchanger Outlet Enthalpy           = {h_sCO2_out_HX/1000:.2f} KJ/kg")
print(f"Heat Exchanger Inlet Entropy             = {s_sCO2_in_HX/1000:.2f} kJ/kg k")
print(f"Heat Exchanger Outlet Entropy            = {s_sCO2_out_HX/1000:.2f} kJ/kg k")

# Turbine Output Statements
print(f"\nTurbine Inlet Temperature                = {T_t_in:.2f} K")
print(f"Turbine Isentropic Outlet Temperature    = {T_t_out_isentropic:.2f} K")
print(f"Turbine Acutal Outlet Temperature        = {T_t_out_actual:.2f} K")
print(f"Turbine Inlet pressure                   = {P_t_in:.2f} Pa")
print(f"Turbine Outlet pressure                  = {P_t_out:.2f} Pa")
print(f"Turbine Specific Volume                  = {T_cv:.2f} J/kg")
print(f"Turbine Specific Heat                    = {T_cp:.2f} J/kg")
print(f"Turbine Gamma                            = {T_gamma:.4f}")
print(f"Turbine Pressure Ratio                   = {T_pressure_ratio:.2f}")
print(f"Turbine Inlet Enthalpy                   = {h_sCO2_in_turbine:.2f} kJ/kg")
print(f"Turbine Outlet Enthalpy                  = {h_sCO2_out_turbine:.2f} kJ/kg")
print(f"Turbine Outlet Isentropic Enthalpy       = {h_sCO2_out_turbine_Isentropic:.2f} kJ/kg")
print(f"Turbine Enthalpy change                  = -{h_Turbine_change:.2f} kJ/kg")
print(f"Turbine Inlet Entropy                    = {s_in_turbine/1000:.2f} kJ/kg")
print(f"Turbine Outlet Entropy                   = {s_out_turbine/1000:.2f} kJ/kg")
print(f"Turbine Outlet Isentropic Entropy        = {s_out_turbine_Isentropic/1000:.2f} kJ/kg")
print(f"Turbine Power Output                     = {Power_Output:.2f} Kw")
print(f"Generator Power Output                   = {Power_Output_e:.2f} Kw")

# Recuperator Output Statements
print(f"\nRecuperator Hot Side Inlet Temperature   = {T_t_out_actual:.2f} K")
print(f"Recuperator Hot Side Outlet Temperature  = {Re_T_hot_out:.2f} K")
print(f"Recuperator Cold Side Inlet Temperature  = {Re_T_cold_in:.2f} K")
print(f"Recuperator Cold Side Outlet Temperature = {Re_T_cold_out:.2f} K")
print(f"Recuperator Hot Side Inlet Pressure      = {P_t_out:.2f} Pa")
print(f"Recuperator Hot Side Outlet Pressure     = {Re_P_hot_out:.2f} Pa")
print(f"Recuperator Cold Side Inlet Pressure     = {Re_P_cold_in:.2f} Pa")
print(f"Recuperator Cold Side Outlet Pressure    = {Re_P_cold_out:.2f} Pa")
print(f"Recuperator Hot side Inlet enthalpy      = {Re_h_hot_in/1000:.2f} kJ/kg")
print(f"Recuperator Hot side Outlet enthalpy     = {Re_h_hot_out/1000:.2f} kJ/kg")
print(f"Recuperator Cold side Inlet enthalpy     = {Re_h_cold_in/1000:.2f} kJ/kg")
print(f"Recuperator Hot side Inlet entropy       = {Re_s_hot_in/1000:.2f} kJ/kg")
print(f"Recuperator Hot side Outlet entropy      = {Re_s_hot_out/1000:.2f} kJ/kg")
print(f"Recuperator Cold side Inlet entropy      = {Re_s_cold_in/1000:.2f} kJ/kg")
print(f"Recuperator Cold side Outlet entropy     = {Re_s_cold_out/1000:.2f} kJ/kg")
print(f"Recuperator Possible Heat transfer       = {Q_dot_recup/1000:.2f} KW")

# Intercooler Output Statements
print(f"\nIntercooler Inlet Temperature            = {T_cooler_in:.2f} K")
print(f"Intercooler Outlet Temperature           = {T_cooler_out:.2f} K")
print(f"Intercooler Ambient Temperature          = {T_cooler_amb:.2f} K")
print(f"Intercooler Inlet  Pressure              = {P_cooler_in:.2f} Pa")
print(f"Intercooler Outlet  Pressure             = {P_cooler_out:.2f} Pa")
print(f"Intercooler Inlet Enthalpy               = {h_cooler_in:.2f} KJ/Kg")
print(f"Intercooler Outlet Enthalpy              = {h_cooler_out:.2f} KJ/Kg")
print(f"Intercooler Inlet Entropy                = {s_cooler_in:.2f} KJ/Kg")
print(f"Intercooler Outlet Entropy               = {s_cooler_out:.2f} KJ/Kg")
print(f"Intercooler Heat Rejection               = {cooler_Q_reject:.2f} KW")

# Compressor Output Statements
print(f"\nCompressor Input Temperature             = {T_comp_in:.2f} K")
print(f"Compressor Output Temperature            = {T_comp_out_actual:.2f} K")
print(f"Compressor Output Isentropic Temperature = {T_comp_out_isentropic:.2f} K")
print(f"Compressor Input Pressure                = {P_comp_in:.2f} Pa")
print(f"Compressor Output Pressure               = {P_comp_out:.2f} Pa")
print(f"Compressor Inlet Enthalpy                = {h_comp_in/1000:.2f} KJ/Kg")
print(f"Compressor Outlet Isentropic Enthalpy    = {h_comp_out_isentropic/1000:.2f} KJ/Kg")
print(f"Compressor Outlet  Enthalpy              = {h_comp_out/1000:.2f} KJ/Kg")
print(f"Compressor Inlet entropy                 = {s_comp_in/1000:.2f} kJ/kg*K")
print(f"Compressor Outlet Isentropic entropy     = {s_comp_out_isentropic/1000:.2f} kJ/kg*K")
print(f"Compressor Outlet entropy                = {s_comp_out/1000:.2f} kJ/kg*K")
print(f"Compressor Power Output                  = {comp_power:.2f} KW")

# Electrical Output
Electric_Out = Power_Output_e - comp_power
print(f"\nNet Electricity Output: {Electric_Out:.2f} KWe")

# Cycle Efficiency
cycle_Efficiency = (Electric_Out * 1e3 / Q) * 100
print(f"Cycle Efficiency: {cycle_Efficiency:.2f}")

# Optional convergence values
print("\nConvergence Check")
print(f"Iterations      = {iter}")
print(f"delta_T         = {delta_T}")
print(f"delta_P         = {delta_P}")
print(f"delta_T_comp    = {delta_T_comp}")
```
